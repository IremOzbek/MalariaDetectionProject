```python
from PIL import Image
import os

def remove_corrupt_images(folder):
    for cls in ["Cat", "Dog"]:
        path = os.path.join(folder, cls)
        for img in os.listdir(path):
            try:
                Image.open(os.path.join(path, img)).verify()
            except:
                os.remove(os.path.join(path, img))

remove_corrupt_images("/content/PetImages")
```

```python
import os
import shutil
import random

SOURCE_DIR = "/content/PetImages"
BASE_DIR = "/content/dataset"

TRAIN_DIR = os.path.join(BASE_DIR, "train")
VAL_DIR = os.path.join(BASE_DIR, "validation")

CATEGORIES = {"Cat": "cats", "Dog": "dogs"}
SPLIT_RATIO = 0.8  # %80 train, %20 validation

# Klasörleri oluştur
for split in [TRAIN_DIR, VAL_DIR]:
    for cls in CATEGORIES.values():
        os.makedirs(os.path.join(split, cls), exist_ok=True)

# Dosyaları ayır
for src_cls, target_cls in CATEGORIES.items():
    images = os.listdir(os.path.join(SOURCE_DIR, src_cls))
    random.shuffle(images)

    split_index = int(len(images) * SPLIT_RATIO)
    train_imgs = images[:split_index]
    val_imgs = images[split_index:]

    for img in train_imgs:
        shutil.copy(
            os.path.join(SOURCE_DIR, src_cls, img),
            os.path.join(TRAIN_DIR, target_cls, img)
        )

    for img in val_imgs:
        shutil.copy(
            os.path.join(SOURCE_DIR, src_cls, img),
            os.path.join(VAL_DIR, target_cls, img)
        )

print("Train / Validation ayrımı tamamlandı.")
```

```python
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator

IMG_SIZE = 224
BATCH_SIZE = 32

train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=20,
    zoom_range=0.2,
    horizontal_flip=True
)

val_datagen = ImageDataGenerator(rescale=1./255)

train_data = train_datagen.flow_from_directory(
    TRAIN_DIR,
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode="binary"
)

val_data = val_datagen.flow_from_directory(
    VAL_DIR,
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode="binary"
)
```

```python
from tensorflow.keras import layers, models

base_model = tf.keras.applications.MobileNetV2(
    input_shape=(IMG_SIZE, IMG_SIZE, 3),
    include_top=False,
    weights="imagenet"
)

base_model.trainable = False  # 🔒 Frozen
```

```python
model = models.Sequential([
    base_model,
    layers.GlobalAveragePooling2D(),
    layers.Dense(128, activation="relu"),
    layers.Dropout(0.3),
    layers.Dense(1, activation="sigmoid")
])
```

```python
model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)
```

```python
history_frozen = model.fit(
    train_data,
    epochs=10,
    validation_data=val_data
)
```

```python
base_model.trainable = True

for layer in base_model.layers[:-30]:
    layer.trainable = False
```

```python
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-5),
    loss="binary_crossentropy",
    metrics=["accuracy"]
)
```

```python
history_finetune = model.fit(
    train_data,
    epochs=5,
    validation_data=val_data
)
```

```python
import matplotlib.pyplot as plt

def plot_history(history, title):
    plt.figure()
    plt.plot(history.history["accuracy"], label="Train Acc")
    plt.plot(history.history["val_accuracy"], label="Val Acc")
    plt.title(title)
    plt.xlabel("Epoch")
    plt.ylabel("Accuracy")
    plt.legend()
    plt.show()

plot_history(history_frozen, "Frozen MobileNetV2")
plot_history(history_finetune, "Fine-Tuned MobileNetV2")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------------
# 1. Validation setinden bir batch al
# -----------------------------------
images, labels = next(val_data)

# -----------------------------------
# 2. Model tahminleri
# -----------------------------------
preds = model.predict(images, verbose=0)

# Sigmoid çıktıyı sınıf etiketine çevir
pred_labels = (preds > 0.5).astype(int).reshape(-1)

# -----------------------------------
# 3. Yanlış tahmin edilen örnekler
# -----------------------------------
wrong_idx = np.where(pred_labels != labels)[0]

print(f"Yanlış sınıflandırılan örnek sayısı: {len(wrong_idx)}")

# -----------------------------------
# 4. Yanlış sınıflandırılan görselleri çiz
# -----------------------------------
num_show = min(8, len(wrong_idx))  # en fazla 8 tane göster

plt.figure(figsize=(14, 6))

for i in range(num_show):
    idx = wrong_idx[i]

    plt.subplot(2, 4, i + 1)
    plt.imshow(images[idx])
    plt.axis("off")

    true_label = "Dog" if labels[idx] == 1 else "Cat"
    pred_label = "Dog" if pred_labels[idx] == 1 else "Cat"
    confidence = preds[idx][0]

    plt.title(
        f"True: {true_label}\nPred: {pred_label}\nConf: {confidence:.2f}",
        fontsize=10,
        color="red"
    )

plt.suptitle("Misclassified Validation Samples", fontsize=14)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

all_images = []
all_labels = []
all_preds = []

# -----------------------------------
# 1. Tüm validation setini dolaş
# -----------------------------------
for i in range(len(val_data)):
    images, labels = val_data[i]
    preds = model.predict(images, verbose=0)

    all_images.extend(images)
    all_labels.extend(labels)
    all_preds.extend(preds)

all_images = np.array(all_images)
all_labels = np.array(all_labels)
all_preds = np.array(all_preds).reshape(-1)

pred_labels = (all_preds > 0.5).astype(int)

# -----------------------------------
# 2. Yanlış sınıflandırılan indeksler
# -----------------------------------
wrong_idx = np.where(pred_labels != all_labels)[0]

print(f"Toplam validation görüntüsü: {len(all_labels)}")
print(f"Yanlış sınıflandırılan görüntü sayısı: {len(wrong_idx)}")

# -----------------------------------
# 3. Yanlış örnekleri görselleştir
# -----------------------------------
num_show = min(8, len(wrong_idx))

plt.figure(figsize=(14, 6))

for i in range(num_show):
    idx = wrong_idx[i]

    plt.subplot(2, 4, i + 1)
    plt.imshow(all_images[idx])
    plt.axis("off")

    true_label = "Dog" if all_labels[idx] == 1 else "Cat"
    pred_label = "Dog" if pred_labels[idx] == 1 else "Cat"
    confidence = all_preds[idx]

    plt.title(
        f"True: {true_label}\nPred: {pred_label}\nConf: {confidence:.2f}",
        fontsize=10,
        color="red"
    )

plt.suptitle("Misclassified Samples from Entire Validation Set", fontsize=14)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix, classification_report, roc_curve, auc
import seaborn as sns
import pandas as pd
```

```python
def plot_acc_loss(history, title):
    epochs = range(1, len(history.history['accuracy']) + 1)

    plt.figure(figsize=(12,4))

    plt.subplot(1,2,1)
    plt.plot(epochs, history.history['accuracy'], label='Train')
    plt.plot(epochs, history.history['val_accuracy'], label='Validation')
    plt.title(f'{title} - Accuracy')
    plt.xlabel('Epoch')
    plt.ylabel('Accuracy')
    plt.legend()

    plt.subplot(1,2,2)
    plt.plot(epochs, history.history['loss'], label='Train')
    plt.plot(epochs, history.history['val_loss'], label='Validation')
    plt.title(f'{title} - Loss')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.legend()

    plt.tight_layout()
    plt.show()

plot_acc_loss(history_frozen, "Frozen MobileNetV2")
plot_acc_loss(history_finetune, "Fine-Tuned MobileNetV2")
```

```python
y_true = []
y_pred = []
y_prob = []

for images, labels in val_data:
    preds = model.predict(images, verbose=0)
    y_true.extend(labels)
    y_prob.extend(preds)
    y_pred.extend((preds > 0.5).astype(int))

y_true = np.array(y_true)
y_pred = np.array(y_pred)
y_prob = np.array(y_prob)
```

```python
cm = confusion_matrix(y_true, y_pred)

plt.figure(figsize=(5,4))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=['Cat', 'Dog'],
            yticklabels=['Cat', 'Dog'])
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.title('Confusion Matrix')
plt.show()
```

```python
fpr, tpr, _ = roc_curve(y_true, y_prob)
roc_auc = auc(fpr, tpr)

plt.figure(figsize=(6,5))
plt.plot(fpr, tpr, label=f'AUC = {roc_auc:.3f}')
plt.plot([0,1], [0,1], linestyle='--')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve')
plt.legend()
plt.show()
```
