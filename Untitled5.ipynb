```python
import tensorflow as tf
import numpy as np

# Test veri setinizin yolunu tanımlayın
# Lütfen test verilerinizi 'MalariaDetectionProject/data/cell_images/test' dizini altına
# (sınıf isimleriyle alt klasörler halinde, örn: Parasitized/, Uninfected/) yükleyin.
TEST_DATA_PATH = 'MalariaDetectionProject/data/cell_images/test'
IMG_SIZE = (128, 128)
BATCH_SIZE = 32

# Test veri setini yükle
try:
    # 'MalariaDetectionProject/data/cell_images/test' klasörü altında en az iki sınıf klasörü (örn: Parasitized, Uninfected) olmalı.
    test_ds = tf.keras.utils.image_dataset_from_directory(
        TEST_DATA_PATH,
        image_size=IMG_SIZE,
        batch_size=BATCH_SIZE,
        shuffle=False # ROC eğrisi için sıralamanın korunması önemlidir
    )

    # Sınıf isimlerini al
    class_names = test_ds.class_names
    print(f"Sınıf isimleri: {class_names}")

    # Test veri setinden görüntüleri ve etiketleri çıkar
    X_test_list = []
    y_test_list = []

    for images, labels in test_ds:
        X_test_list.append(images.numpy())
        y_test_list.append(labels.numpy())

    X_test = np.concatenate(X_test_list, axis=0)
    y_test = np.concatenate(y_test_list, axis=0)

    # Görüntü piksellerini 0-1 aralığına ölçekle
    X_test = X_test / 255.0

    print(f"Test veri seti başarıyla yüklendi. Boyut: {X_test.shape}, Etiketler: {y_test.shape}")

except Exception as e:
    print(f"Test veri seti yüklenirken bir hata oluştu: {e}")
    print(f"Lütfen test verilerinizi '{TEST_DATA_PATH}' dizinine yüklediğinizden ve")
    print("test veri dizininin, alt klasörlerde sınıflandırılmış görüntüler (örn: Parasitized/, Uninfected/) içerdiğinden emin olun.")
```

Yukarıdaki kodu çalıştırdıktan sonra, eğer test verileriniz başarıyla yüklenirse, `X_test` ve `y_test` değişkenleri ROC eğrisini oluşturmak için hazır olacaktır. Ardından, daha önceki hücredeki ROC eğrisi oluşturma kodunu `X_test` ve `y_test` değişkenlerini kullanarak devam ettirebilirsiniz. `YOUR_ACTUAL_TEST_DATA_PATH` kısmını kendi test veri setinizin yolu ile değiştirmeyi unutmayın.

```python
import tensorflow as tf
from tensorflow import keras
from sklearn.metrics import roc_curve, auc
import numpy as np
import matplotlib.pyplot as plt
import os

# Projenin src klasörünü path'e ekle
import sys
sys.path.append('MalariaDetectionProject/src')
from data_loader import get_datasets

# Model dosya yolu
model_path = 'MalariaDetectionProject/models/best_malaria_model.keras'

# 1. Modeli Yükle
try:
    model = keras.models.load_model(model_path)
    print(f'Model başarıyla yüklendi: {model_path}')
except Exception as e:
    print(f'Model yüklenirken bir hata oluştu: {e}')
    exit()

# 2. Doğrulama Verilerini Yükle ve Hazırla (Test Verisi Yerine Kullanılacak)
# Mevcut veri setinizi %80 eğitim, %20 doğrulama olarak ayırdığınız için
# doğrulama setini ROC eğrisi için kullanacağız.

data_path = 'MalariaDetectionProject/data/cell_images'
_, val_ds, _ = get_datasets(data_path) # Eğitim setine ihtiyacımız yok

print(f'Doğrulama veri seti başarıyla yüklendi.')

# val_ds'ten görüntüleri ve etiketleri çıkar
X_val_list = []
y_val_list = []

for images, labels in val_ds:
    X_val_list.append(images.numpy())
    y_val_list.append(labels.numpy())

X_val = np.concatenate(X_val_list, axis=0)
y_val = np.concatenate(y_val_list, axis=0)

# Görüntü piksellerini 0-1 aralığına ölçekle (get_datasets içinde yapılmıyorsa)
X_val = X_val / 255.0

print(f'Doğrulama veri seti boyutu: {len(X_val)} örnek')

# 3. Model ile Tahminler Yap
y_pred_proba = model.predict(X_val)

# 4. ROC Eğrisini ve AUC Değerini Hesapla
fpr, tpr, thresholds = roc_curve(y_val, y_pred_proba)
roc_auc = auc(fpr, tpr)

# 5. ROC Eğrisini Çizdir
plt.figure(figsize=(8, 6))
plt.plot(fpr, tpr, color='darkorange', lw=2, label='ROC curve (area = %0.2f)' % roc_auc)
plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--')
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('Receiver Operating Characteristic (ROC) Curve')
plt.legend(loc='lower right')
plt.grid(True)
plt.show()

print(f'AUC Skoru: {roc_auc:.4f}')
```

```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.models import load_model
from sklearn.metrics import roc_curve, auc
import sys

# Add the project's src directory to the Python path
sys.path.append('./MalariaDetectionProject/src')

from data_loader import get_datasets  # Kendi yazdığın data_loader'ı kullanıyoruz

def plot_roc_curve(model_path, data_path):
    # 1. Veri setini ve Modeli Yükle
    # img_size ve batch_size eğitimde kullandığın değerlerle aynı olmalı
    _, val_ds, _ = get_datasets(data_path, img_size=(128, 128), batch_size=32)
    model = load_model(model_path)

    print("Tahminler yapılıyor, lütfen bekleyin...")

    y_true = []
    y_probs = []

    # 2. Doğrulama setindeki tüm resimler için olasılıkları topla
    for images, labels in val_ds:
        preds = model.predict(images, verbose=0)
        y_true.extend(labels.numpy())
        y_probs.extend(preds.ravel())

    y_true = np.array(y_true)
    y_probs = np.array(y_probs)

    # 3. ROC değerlerini hesapla
    fpr, tpr, thresholds = roc_curve(y_true, y_probs)
    roc_auc = auc(fpr, tpr)

    # 4. Çizim Yap
    plt.figure(figsize=(10, 7))
    plt.plot(fpr, tpr, color='#2980b9', lw=3, label=f'ROC Eğrisi (AUC = {roc_auc:.4f})')
    plt.plot([0, 1], [0, 1], color='#c0392b', lw=2, linestyle='--', label='Rastgele Tahmin')

    plt.xlim([0.0, 1.0])
    plt.ylim([0.0, 1.05])
    plt.xlabel('Yanlış Pozitif Oranı (False Positive Rate)', fontsize=12)
    plt.ylabel('Doğru Pozitif Oranı (True Positive Rate)', fontsize=12)
    plt.title('Sıtma Teşhis Modeli ROC Analizi', fontsize=14)
    plt.legend(loc="lower right")
    plt.grid(alpha=0.3)

    # Grafiği kaydet ve göster
    plt.savefig("reports/roc_curve_analysis.png")
    plt.show()
    print(f"ROC eğrisi kaydedildi. AUC Skoru: {roc_auc:.4f}")

if __name__ == "__main__":
    # Dosya yollarını projene göre güncelle
    MODEL_PATH = "MalariaDetectionProject/models/best_malaria_model.keras"
    DATA_PATH = "MalariaDetectionProject/data/cell_images"

    plot_roc_curve(MODEL_PATH, DATA_PATH)
```

```python
import os
import sys
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from sklearn.metrics import roc_curve, auc

# 1. Proje dizinini Python yoluna ekle
project_path = os.path.abspath("/content/MalariaDetectionProject")
if project_path not in sys.path:
    sys.path.append(project_path)

# Standart veri yükleyiciyi içe aktar (Wavelet içermeyen versiyon)
from src.data_loader import get_datasets

def generate_standard_cnn_roc():
    # Yapılandırma - Standart Model Yolları
    model_path = "models/best_malaria_model.keras"
    data_path = "data/cell_images"
    report_path = "reports/standard_cnn_roc_analysis.png"

    print("\n" + "="*50)
    print("🚀 STANDART CNN ROC ANALİZİ BAŞLATILDI")
    print("="*50)

    # 2. Modeli Yükle
    if not os.path.exists(model_path):
        print(f"❌ HATA: '{model_path}' bulunamadı!")
        return

    print(f"📦 Standart model yükleniyor: {model_path}")
    model = tf.keras.models.load_model(model_path)

    # 3. Veriyi Yükle (Orijinal 128x128x3 boyutunda)
    print("🔄 Standart doğrulama veri seti hazırlanıyor...")
    # get_datasets fonksiyonun orijinal halindeyse (Wavelet yoksa) direkt çağır
    _, val_ds, _ = get_datasets(data_path, img_size=(128, 128), batch_size=32)

    # 4. Tahminleri Topla
    print("🧠 Tahminler yapılıyor...")
    y_true = []
    y_probs = []

    for images, labels in val_ds:
        preds = model.predict(images, verbose=0)
        y_true.extend(labels.numpy())
        y_probs.extend(preds.ravel())

    y_true = np.array(y_true)
    y_probs = np.array(y_probs)

    # 5. ROC ve AUC Hesaplama
    fpr, tpr, _ = roc_curve(y_true, y_probs)
    roc_auc = auc(fpr, tpr)

    # 6. Görselleştirme
    print("📊 Grafik oluşturuluyor...")
    plt.figure(figsize=(10, 8))
    plt.plot(fpr, tpr, color='#e67e22', lw=3, label=f'Standart CNN (AUC = {roc_auc:.4f})')
    plt.plot([0, 1], [0, 1], color='#2c3e50', lw=2, linestyle='--', label='Şans Oranı')

    plt.fill_between(fpr, tpr, alpha=0.1, color='#e67e22')
    plt.xlim([0.0, 1.0])
    plt.ylim([0.0, 1.05])
    plt.xlabel('Yanlış Pozitif Oranı (FPR)', fontsize=12)
    plt.ylabel('Doğru Pozitif Oranı (TPR)', fontsize=12)
    plt.title('Standart CNN Sıtma Tespit: ROC Eğrisi', fontsize=14)
    plt.legend(loc="lower right", fontsize=11)
    plt.grid(True, linestyle=':', alpha=0.6)

    # Kaydet ve Göster
    os.makedirs("reports", exist_ok=True)
    plt.savefig(report_path)
    plt.show()

    print("\n" + "="*50)
    print(f"✅ STANDART MODEL ANALİZİ TAMAMLANDI!")
    print(f"📈 AUC Skoru: {roc_auc:.4f}")
    print(f"📂 Grafik kaydedildi: {report_path}")
    print("="*50)

if __name__ == "__main__":
    generate_standard_cnn_roc()
```
